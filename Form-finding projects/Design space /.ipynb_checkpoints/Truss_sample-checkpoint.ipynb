```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

def plot_fdm(grid_points, new_node_coords, connectivity, Free_nodes=None):
    """
    Plot comparison between original grid points and deformed FDM solution using matplotlib
    
    Parameters:
    - grid_points: Tensor/array of original node coordinates (Nx3)
    - new_node_coords: Tensor/array of deformed node coordinates (Nx3)
    - connectivity: List of node connections (element edges) as pairs [i,j]
    - Free_nodes: Optional list of free nodes to highlight (default None)
    """

    grid_points = torch.tensor(grid_points)
    # Convert to numpy if they're tensors
    if hasattr(grid_points, 'numpy'):
        grid_points = grid_points.numpy()
    if hasattr(new_node_coords, 'numpy'):
        new_node_coords = new_node_coords.numpy()
    
    fig = plt.figure(figsize=(12, 8))
    ax = fig.add_subplot(111, projection='3d')
    
    # Plot original structure (blue)
    for i, j in connectivity:
        ax.plot([grid_points[i-1, 0], grid_points[j-1, 0]],
                [grid_points[i-1, 1], grid_points[j-1, 1]],
                [grid_points[i-1, 2], grid_points[j-1, 2]],
                'b--', linewidth=2, alpha=0.7, label='Original' if (i,j) == connectivity[0] else "")
    
    # Plot deformed structure (red)
    for i, j in connectivity:
        ax.plot([new_node_coords[i-1, 0], new_node_coords[j-1, 0]],
                [new_node_coords[i-1, 1], new_node_coords[j-1, 1]],
                [new_node_coords[i-1, 2], new_node_coords[j-1, 2]],
                'r-', linewidth=3, label='FDM Solution' if (i,j) == connectivity[0] else "")
    
    
    # Calculate axis ranges with padding
    all_coords = np.vstack([grid_points, new_node_coords])
    min_vals = all_coords.min(axis=0)
    max_vals = all_coords.max(axis=0)
    ranges = max_vals - min_vals
    padding = 0.1 * ranges
    
    ax.set_xlim([min_vals[0]-padding[0], max_vals[0]+padding[0]])
    ax.set_ylim([min_vals[1]-padding[1], max_vals[1]+padding[1]])
    ax.set_zlim([min_vals[2]-padding[2], max_vals[2]+padding[2]])
    
    ax.set_xlabel('Y Axis')
    ax.set_ylabel('X Axis')
    ax.set_zlabel('Z Axis')
    ax.set_title('FDM Geometry')
    
    # Create custom legend without duplicates
    handles, labels = ax.get_legend_handles_labels()
    by_label = dict(zip(labels, handles))  # Remove duplicates
    ax.legend(by_label.values(), by_label.keys(), loc='upper right')
    
    plt.tight_layout()
    return fig
```

```python
########## Differentiable FDM
Fixed_nodes = torch.tensor([1,3,4])
Free_nodes = torch.tensor([2])
idx_CF = Fixed_nodes - 1
idx_CN = Free_nodes - 1
f_n = idx_CN
px = torch.zeros(len(f_n), 1, dtype=torch.float32)
py = torch.zeros(len(f_n), 1, dtype=torch.float32)
pz = torch.zeros(len(f_n), 1, dtype=torch.float32)

pz[:, 0] = 1.0
print(pz)
## Obtaining Matrix D
# Assembly the nodal-branch matrix:
n_elements = len(connectivity)
n_nodes = len(idx_CF) + len(idx_CN)
C = torch.zeros(n_elements, n_nodes, dtype=torch.float32)
for n, (i, j) in enumerate(connectivity):
    C[n, i - 1] = 1
    C[n, j - 1] = -1
CF = C[:, idx_CF]
CN = C[:, idx_CN]


# Assembly D
def FDM (Q, C=C, CN=CN, CF=CF, px=px, py=py, pz=pz,Fixed_nodes=Fixed_nodes, Free_nodes=Free_nodes,node_coords=grid_points) :
    Dn = torch.matmul(torch.transpose(CN, 0, 1), torch.matmul(Q, CN))
    DF = torch.matmul(torch.transpose(CN, 0, 1), torch.matmul(Q, CF))
    ## Coordinates solution
    # separate the coordinates:
    node_coords = torch.tensor(node_coords) 
    fixed_idces = [node - 1 for node in Fixed_nodes]
    xF = node_coords[fixed_idces, 0].unsqueeze(1)
    yF = node_coords[fixed_idces, 1].unsqueeze(1)
    zF = node_coords[fixed_idces, 2].unsqueeze(1)
    xN = torch.matmul(torch.inverse(Dn), (px - torch.matmul(DF, xF)))
    yN = torch.matmul(torch.inverse(Dn), (py - torch.matmul(DF, yF)))
    zN = torch.matmul(torch.inverse(Dn), (pz - torch.matmul(DF, zF)))
    ## Coordinates renewal: 
    new_node_coords = node_coords.clone()
    for idx, node in enumerate(Free_nodes):
        node_idx = node - 1
        new_node_coords[node_idx, 0] = xN[idx]
        new_node_coords[node_idx, 1] = yN[idx]
        new_node_coords[node_idx, 2] = zN[idx]
    return new_node_coords





####### BCs
fixed_dof = []
for node in Fixed_nodes:
    fixed_dof.extend([(node - 1) * 6 + i for i in range(6)])
    
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.patches import Rectangle

# 设置设备
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# 常量定义
D_radius = 0.75
D_young_modulus = 10e9 
D_shear_modulus = 0.7e9 
D_poisson_ratio = 0.3
cross_section_angle_a = 0  
cross_section_angle_b = 0  
a_small_number = 1e-10

# 旋转函数
def rotation(v, k, theta):
    k = k / torch.norm(k)  # 归一化
    cross_product = torch.cross(k, v)
    dot_product = torch.dot(k, v)
    theta = torch.tensor(theta, dtype=torch.float32, device=device) if not isinstance(theta, torch.Tensor) else theta
    v_rotated = v * torch.cos(theta) + cross_product * torch.sin(theta) + k * dot_product * (1 - torch.cos(theta))
    return v_rotated

# 梁单元类
class Beam:
    def __init__(self, node_coordinates, R=D_radius, young_modulus=D_young_modulus,
                 shear_modulus=D_shear_modulus, poisson_ratio=D_poisson_ratio, Beta_a=cross_section_angle_a,
                 Beta_b=cross_section_angle_b):
        self.node_coordinates = node_coordinates

        # 材料和几何属性
        self.radius = R
        self.young_modulus = young_modulus
        self.shear_modulus = shear_modulus
        self.poisson_ratio = poisson_ratio

        # 截面属性
        self.length = torch.norm(self.node_coordinates[1] - self.node_coordinates[0])
        self.Iy = (torch.pi * self.radius ** 4) / 4 
        self.Iz = self.Iy
        self.A = torch.pi * self.radius ** 2
        self.J = (torch.pi * self.radius ** 4) / 2

        # 刚度分量
        self.S_u = self.young_modulus * self.A / self.length
        self.S_v1a = 12 * self.young_modulus * self.Iy / (self.length ** 3)
        self.S_v1b = 6 * self.young_modulus * self.Iy / (self.length ** 2)
        self.S_v2a = 12 * self.young_modulus * self.Iz / (self.length ** 3)
        self.S_v2b = 6 * self.young_modulus * self.Iz / (self.length ** 2)
        self.S_theta1a = 6 * self.young_modulus * self.Iy / (self.length ** 2)
        self.S_theta1b = 4 * self.young_modulus * self.Iy / self.length
        self.S_theta1c = 2 * self.young_modulus * self.Iy / self.length
        self.S_theta2a = 6 * self.young_modulus * self.Iz / (self.length ** 2)
        self.S_theta2b = 4 * self.young_modulus * self.Iz / self.length
        self.S_theta2c = 2 * self.young_modulus * self.Iz / self.length
        self.S_Tr = self.shear_modulus * self.J / self.length

        # 截面旋转
        self.Beta_a = Beta_a
        self.Beta_b = Beta_b

    def get_element_stiffness_matrix(self):
        K_element = torch.tensor([
            [self.S_u, 0, 0, 0, 0, 0, -self.S_u, 0, 0, 0, 0, 0],
            [0, self.S_v1a, 0, 0, 0, self.S_theta1a, 0, -self.S_v1a, 0, 0, 0, self.S_theta1a],
            [0, 0, self.S_v2a, 0, -self.S_theta2a, 0, 0, 0, -self.S_v2a, 0, -self.S_theta2a, 0],
            [0, 0, 0, self.S_Tr, 0, 0, 0, 0, 0, -self.S_Tr, 0, 0],
            [0, 0, -self.S_v2b, 0, self.S_theta2b, 0, 0, 0, self.S_v2b, 0, self.S_theta2c, 0],
            [0, self.S_v1b, 0, 0, 0, self.S_theta1b, 0, -self.S_v1b, 0, 0, 0, self.S_theta1c],
            [-self.S_u, 0, 0, 0, 0, 0, self.S_u, 0, 0, 0, 0, 0],
            [0, -self.S_v1a, 0, 0, 0, -self.S_theta1a, 0, self.S_v1a, 0, 0, 0, -self.S_theta1a],
            [0, 0, -self.S_v2a, 0, self.S_theta2a, 0, 0, 0, self.S_v2a, 0, self.S_theta2a, 0],
            [0, 0, 0, -self.S_Tr, 0, 0, 0, 0, 0, self.S_Tr, 0, 0],
            [0, 0, -self.S_v2b, 0, self.S_theta2c, 0, 0, 0, self.S_v2b, 0, self.S_theta2b, 0],
            [0, self.S_v1b, 0, 0, 0, self.S_theta1c, 0, -self.S_v1b, 0, 0, 0, self.S_theta1b],
        ], dtype=torch.float32, device=device)

        return K_element

    def System_Transform(self):
        vector_x = self.node_coordinates[1, 0] - self.node_coordinates[0, 0]
        vector_y = self.node_coordinates[1, 1] - self.node_coordinates[0, 1]
        vector_z = self.node_coordinates[1, 2] - self.node_coordinates[0, 2]
        length = torch.norm(self.node_coordinates[1] - self.node_coordinates[0])
        
        z_value = torch.clamp(vector_z / length, min=-1 + 1e-6, max=1 - 1e-6)
        ceta = torch.acos(z_value)
        value = vector_x / torch.sqrt(vector_y ** 2 + vector_x ** 2 + a_small_number)
        value = torch.clamp(value, min=-1 + 1e-6, max=1 - 1e-6)
        alpha = torch.acos(value)

        Projection_Z_x = - vector_z / length * torch.sin(alpha)
        Projection_Z_y = - vector_z / length * torch.cos(alpha)
        Projection_Z_z = torch.cos(torch.pi / 2 - ceta)

        V_projection = torch.stack([Projection_Z_x, Projection_Z_y, Projection_Z_z])
        X_axis = torch.stack([vector_x / length, vector_y / length, vector_z / length])
        Z_axis_a = rotation(V_projection, X_axis, self.Beta_a)
        Y_axis_a = rotation(Z_axis_a, X_axis, -torch.pi / 2)
        Z_axis_a = Z_axis_a / torch.norm(Z_axis_a)
        Y_axis_a = Y_axis_a / torch.norm(Y_axis_a)

        lambda_matrix = torch.stack([X_axis, Y_axis_a, Z_axis_a], dim=0)
        matrix_T = torch.zeros((12, 12), dtype=torch.float32, device=device)
        for i in range(0, 12, 3):
            matrix_T[i:i + 3, i:i + 3] = lambda_matrix
        return matrix_T

# 组装全局刚度矩阵
def assemble_stiffness_matrix(beams, n_nodes, n_dof_per_node, connectivity):
    total_dof = n_nodes * n_dof_per_node
    K_global = torch.zeros((total_dof, total_dof), dtype=torch.float32, device=device)
    
    for idx, (i, j) in enumerate(connectivity):
        Matrix_T = beams[idx].System_Transform()
        K_element = torch.matmul(torch.transpose(Matrix_T, 0, 1),
                                 torch.matmul(beams[idx].get_element_stiffness_matrix(), Matrix_T))

        start_idx = (i - 1) * n_dof_per_node
        end_idx = (j - 1) * n_dof_per_node
        K_global[start_idx:start_idx + 6, start_idx:start_idx + 6] += K_element[0:6, 0:6]
        K_global[end_idx:end_idx + 6, end_idx:end_idx + 6] += K_element[6:12, 6:12]
        K_global[start_idx:start_idx + 6, end_idx:end_idx + 6] += K_element[0:6, 6:12]
        K_global[end_idx:end_idx + 6, start_idx:start_idx + 6] += K_element[6:12, 0:6]

    return K_global

# 鲁棒求解器
def robust_solve(K_global, F, fixed_dof, max_attempts=3):
    attempts = 0
    while attempts < max_attempts:
        reg = 1e-6 * torch.eye(K_global.shape[0], device=K_global.device)
        reg[fixed_dof, fixed_dof] = 0
        K_reg = K_global + reg
        
        try:
            displacements = torch.linalg.solve(
                K_reg.to(torch.float64), 
                F.to(torch.float64)
            )
            return displacements.to(K_global.dtype)
        except RuntimeError:
            diag = torch.diag(K_global)
            extreme_mask = (diag > 1e12) & (~torch.isin(torch.arange(len(diag)), torch.tensor(fixed_dof)))  
            K_reg[extreme_mask] = 0
            K_reg[:, extreme_mask] = 0
            K_reg[extreme_mask, extreme_mask] = 1e12
            
            K_reg[fixed_dof, :] = 0
            K_reg[:, fixed_dof] = 0
            K_reg[fixed_dof, fixed_dof] = 1e10
            
            try:
                displacements, info = torch.linalg.cg(
                    K_reg.to(torch.float64),
                    F.to(torch.float64),
                    maxiter=5000,
                    atol=1e-6
                )
                if info > 0:
                    raise RuntimeError("CG failed")
                return displacements.to(K_global.dtype)
            except:
                K_pinv = torch.linalg.pinv(K_reg)
                K_pinv[fixed_dof, :] = 0
                displacements = K_pinv @ F
                print("Warning: Used pseudo-inverse, accuracy may be reduced")
                return displacements
        attempts += 1
    
    raise RuntimeError("Failed to solve linear system")

# 应变能计算
def calculate_strain_energy(node_coords, connectivity, fixed_dof, F):
    Beam_lens = []
    beams = []
    for connection in connectivity:
        node_1_coords = node_coords[connection[0] - 1]
        node_2_coords = node_coords[connection[1] - 1]
        beam = Beam(node_coordinates=torch.stack([node_1_coords, node_2_coords]),
                    R=D_radius, young_modulus=D_young_modulus,
                    shear_modulus=D_shear_modulus, poisson_ratio=D_poisson_ratio, 
                    Beta_a=cross_section_angle_a, Beta_b=cross_section_angle_b)
        beams.append(beam)
        Beam_lens.append(beam.length)
    
    K_global = assemble_stiffness_matrix(beams, n_nodes=len(node_coords), 
                                        n_dof_per_node=6, connectivity=connectivity)
    K_global[fixed_dof, :] = 0
    K_global[:, fixed_dof] = 0
    K_global[fixed_dof, fixed_dof] = 1e10

    displacements = robust_solve(K_global, F, fixed_dof)

    strain_energy_list = []
    ASE_list = []
    Local_d = torch.zeros(len(connectivity), 12, dtype=torch.float32, device=device)
    
    for n, (i, j) in enumerate(connectivity):
        matrix_T = beams[n].System_Transform()
        Tep_displacements = torch.cat(
            [displacements[6 * (i - 1):6 * (i - 1) + 6], 
             displacements[6 * (j - 1):6 * (j - 1) + 6]], dim=0)
        Local_d_n = torch.matmul(Tep_displacements, matrix_T.T)
        Local_d[n, :] = Local_d_n.clone()
        K_l = beams[n].get_element_stiffness_matrix()
        
        # 总应变能
        strain_energy = 0.5 * torch.matmul(Local_d_n, torch.matmul(K_l, Local_d_n.reshape(-1, 1)))
        strain_energy_list.append(strain_energy)
        
        # 轴向应变能 (ASE)
        axial_energy = 0.5 * (Local_d_n[0]-Local_d_n[6]) * beams[n].S_u * (Local_d_n[0]-Local_d_n[6])
        ASE_list.append(axial_energy)
    
    Strain_energy = torch.stack(strain_energy_list)
    ASE = torch.stack(ASE_list)
    Ratio = torch.sum(ASE) / torch.sum(Strain_energy)
    
    return Ratio.item()

# 主函数：计算并绘制Ratio分布
def main():
    # 采样设置
    resolution = 30  # 网格分辨率
    z2_values = np.linspace(0, 2, resolution)
    z3_values = np.linspace(0, 2, resolution)
    ratio_grid = np.zeros((resolution, resolution))
    
    # 固定节点
    fixed_nodes = [0, 3]  # 节点1和节点4（索引从0开始）
    
    # 连接关系
    connectivity = [(1, 2), (2, 3), (3, 4)]  # 杆件连接
    
    # 荷载设置（在节点2和3施加垂直荷载）
    F = torch.zeros(24, dtype=torch.float32, device=device)  # 4节点 * 6自由度
    F[6*1 + 2] = -1000  # 节点2的z方向力 (索引: 1*6 + 2 = 8)
    F[6*2 + 2] = -1000  # 节点3的z方向力 (索引: 2*6 + 2 = 14)
    
    # 固定自由度（节点1和节点4的所有自由度）
    fixed_dof = []
    for node_idx in fixed_nodes:
        fixed_dof.extend(range(node_idx*6, node_idx*6+6))
    
    # 进度计数器
    total_points = resolution * resolution
    completed = 0
    
    # 遍历所有采样点
    for i, z2 in enumerate(z2_values):
        for j, z3 in enumerate(z3_values):
            # 节点坐标 (x, y=0, z)
            node_coords = torch.tensor([
                [0.0, 0.0, 0.0],   # 节点1
                [2.0, 0.0, z2],    # 节点2
                [4.0, 0.0, z3],    # 节点3
                [6.0, 0.0, 0.0]    # 节点4
            ], dtype=torch.float32, device=device)
            
            try:
                ratio = calculate_strain_energy(node_coords, connectivity, fixed_dof, F)
                ratio_grid[j, i] = ratio  # j对应y轴(z3)，i对应x轴(z2)
            except Exception as e:
                print(f"Error at (z2={z2:.2f}, z3={z3:.2f}): {str(e)}")
                ratio_grid[j, i] = np.nan
            
            completed += 1
            if completed % 50 == 0:
                print(f"Progress: {completed}/{total_points} ({completed/total_points*100:.1f}%)")
    
    # 绘制结果
    plt.figure(figsize=(12, 10))
    
    # 创建自定义颜色映射
    colors = [(0.8, 0.0, 0.0), (1.0, 0.8, 0.0), (0.0, 0.6, 0.0)]  # 红 -> 黄 -> 绿
    cmap = LinearSegmentedColormap.from_list("custom", colors, N=256)
    
    # 绘制应变能比率热图
    im = plt.imshow(ratio_grid, extent=[0, 2, 0, 2], origin='lower', 
                   cmap=cmap, vmin=0, vmax=1)
    plt.colorbar(im, label='Axial Strain Energy Ratio (ASE/Total)')
    
    # 标记高比率区域 (Ratio > 0.8)
    high_ratio_mask = ratio_grid > 0.8
    z2_mesh, z3_mesh = np.meshgrid(z2_values, z3_values)
    plt.contour(z2_mesh, z3_mesh, high_ratio_mask.astype(float), levels=[0.5], 
                colors='blue', linewidths=2)
    
    # 标记抛物线解
    parabola_line = np.linspace(0, 2, 100)
    plt.plot(parabola_line, parabola_line, 'r--', linewidth=2, label='Parabolic Solution (z2=z3)')
    
    # 添加标记点
    plt.plot(1.0, 1.0, 'ro', markersize=8, label='Parabolic Point (1.0, 1.0)')
    plt.plot(0.0, 0.0, 'kx', markersize=8, label='Flat Structure (0.0, 0.0)')
    
    # 设置标题和标签
    plt.title('Axial Strain Energy Ratio Distribution\n(z2 and z3 in [0, 2] range)', fontsize=14)
    plt.xlabel('Node 2 z-coordinate (z2)', fontsize=12)
    plt.ylabel('Node 3 z-coordinate (z3)', fontsize=12)
    
    # 添加网格和图例
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.legend(loc='upper left')
    
    # 添加说明文本框
    props = dict(boxstyle='round', facecolor='white', alpha=0.8)
    textstr = '\n'.join((
        'Structure Parameters:',
        f'- Young\'s modulus: {D_young_modulus/1e9:.1f} GPa',
        f'- Radius: {D_radius:.2f} m',
        f'- Load: 1000N downward at nodes 2 & 3',
        f'- Fixed nodes: 1 & 4'))
    plt.text(1.05, 0.05, textstr, transform=plt.gca().transAxes, 
             fontsize=10, verticalalignment='bottom', bbox=props)
    
    plt.tight_layout()
    plt.savefig('strain_energy_ratio_distribution.png', dpi=300)
    plt.show()
    
    # 打印高比率区域统计数据
    high_ratio_points = np.sum(high_ratio_mask)
    total_points = resolution * resolution
    print(f"\nResults Summary:")
    print(f"Total points: {total_points}")
    print(f"Points with Ratio > 0.8: {high_ratio_points} ({high_ratio_points/total_points*100:.1f}%)")
    print(f"Minimum Ratio: {np.nanmin(ratio_grid):.4f}")
    print(f"Maximum Ratio: {np.nanmax(ratio_grid):.4f}")
    print(f"Average Ratio: {np.nanmean(ratio_grid):.4f}")

if __name__ == "__main__":
    main()
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

# 设置设备
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# 常量定义
D_radius = 0.75
D_young_modulus = 10e9 
D_shear_modulus = 0.7e9 
D_poisson_ratio = 0.3
cross_section_angle_a = 0  
cross_section_angle_b = 0  
a_small_number = 1e-10

# 旋转函数
def rotation(v, k, theta):
    k = k / torch.norm(k)  # 归一化
    cross_product = torch.cross(k, v)
    dot_product = torch.dot(k, v)
    theta = torch.tensor(theta, dtype=torch.float32, device=device) if not isinstance(theta, torch.Tensor) else theta
    v_rotated = v * torch.cos(theta) + cross_product * torch.sin(theta) + k * dot_product * (1 - torch.cos(theta))
    return v_rotated

# 梁单元类
class Beam:
    def __init__(self, node_coordinates, R=D_radius, young_modulus=D_young_modulus,
                 shear_modulus=D_shear_modulus, poisson_ratio=D_poisson_ratio, Beta_a=cross_section_angle_a,
                 Beta_b=cross_section_angle_b):
        self.node_coordinates = node_coordinates

        # 材料和几何属性
        self.radius = R
        self.young_modulus = young_modulus
        self.shear_modulus = shear_modulus
        self.poisson_ratio = poisson_ratio

        # 截面属性
        self.length = torch.norm(self.node_coordinates[1] - self.node_coordinates[0])
        self.Iy = (torch.pi * self.radius ** 4) / 4 
        self.Iz = self.Iy
        self.A = torch.pi * self.radius ** 2
        self.J = (torch.pi * self.radius ** 4) / 2

        # 刚度分量
        self.S_u = self.young_modulus * self.A / self.length
        self.S_v1a = 12 * self.young_modulus * self.Iy / (self.length ** 3)
        self.S_v1b = 6 * self.young_modulus * self.Iy / (self.length ** 2)
        self.S_v2a = 12 * self.young_modulus * self.Iz / (self.length ** 3)
        self.S_v2b = 6 * self.young_modulus * self.Iz / (self.length ** 2)
        self.S_theta1a = 6 * self.young_modulus * self.Iy / (self.length ** 2)
        self.S_theta1b = 4 * self.young_modulus * self.Iy / self.length
        self.S_theta1c = 2 * self.young_modulus * self.Iy / self.length
        self.S_theta2a = 6 * self.young_modulus * self.Iz / (self.length ** 2)
        self.S_theta2b = 4 * self.young_modulus * self.Iz / self.length
        self.S_theta2c = 2 * self.young_modulus * self.Iz / self.length
        self.S_Tr = self.shear_modulus * self.J / self.length

        # 截面旋转
        self.Beta_a = Beta_a
        self.Beta_b = Beta_b

    def get_element_stiffness_matrix(self):
        K_element = torch.tensor([
            [self.S_u, 0, 0, 0, 0, 0, -self.S_u, 0, 0, 0, 0, 0],
            [0, self.S_v1a, 0, 0, 0, self.S_theta1a, 0, -self.S_v1a, 0, 0, 0, self.S_theta1a],
            [0, 0, self.S_v2a, 0, -self.S_theta2a, 0, 0, 0, -self.S_v2a, 0, -self.S_theta2a, 0],
            [0, 0, 0, self.S_Tr, 0, 0, 0, 0, 0, -self.S_Tr, 0, 0],
            [0, 0, -self.S_v2b, 0, self.S_theta2b, 0, 0, 0, self.S_v2b, 0, self.S_theta2c, 0],
            [0, self.S_v1b, 0, 0, 0, self.S_theta1b, 0, -self.S_v1b, 0, 0, 0, self.S_theta1c],
            [-self.S_u, 0, 0, 0, 0, 0, self.S_u, 0, 0, 0, 0, 0],
            [0, -self.S_v1a, 0, 0, 0, -self.S_theta1a, 0, self.S_v1a, 0, 0, 0, -self.S_theta1a],
            [0, 0, -self.S_v2a, 0, self.S_theta2a, 0, 0, 0, self.S_v2a, 0, self.S_theta2a, 0],
            [0, 0, 0, -self.S_Tr, 0, 0, 0, 0, 0, self.S_Tr, 0, 0],
            [0, 0, -self.S_v2b, 0, self.S_theta2c, 0, 0, 0, self.S_v2b, 0, self.S_theta2b, 0],
            [0, self.S_v1b, 0, 0, 0, self.S_theta1c, 0, -self.S_v1b, 0, 0, 0, self.S_theta1b],
        ], dtype=torch.float32, device=device)

        return K_element

    def System_Transform(self):
        vector_x = self.node_coordinates[1, 0] - self.node_coordinates[0, 0]
        vector_y = self.node_coordinates[1, 1] - self.node_coordinates[0, 1]
        vector_z = self.node_coordinates[1, 2] - self.node_coordinates[0, 2]
        length = torch.norm(self.node_coordinates[1] - self.node_coordinates[0])
        
        z_value = torch.clamp(vector_z / length, min=-1 + 1e-6, max=1 - 1e-6)
        ceta = torch.acos(z_value)
        value = vector_x / torch.sqrt(vector_y ** 2 + vector_x ** 2 + a_small_number)
        value = torch.clamp(value, min=-1 + 1e-6, max=1 - 1e-6)
        alpha = torch.acos(value)

        Projection_Z_x = - vector_z / length * torch.sin(alpha)
        Projection_Z_y = - vector_z / length * torch.cos(alpha)
        Projection_Z_z = torch.cos(torch.pi / 2 - ceta)

        V_projection = torch.stack([Projection_Z_x, Projection_Z_y, Projection_Z_z])
        X_axis = torch.stack([vector_x / length, vector_y / length, vector_z / length])
        Z_axis_a = rotation(V_projection, X_axis, self.Beta_a)
        Y_axis_a = rotation(Z_axis_a, X_axis, -torch.pi / 2)
        Z_axis_a = Z_axis_a / torch.norm(Z_axis_a)
        Y_axis_a = Y_axis_a / torch.norm(Y_axis_a)

        lambda_matrix = torch.stack([X_axis, Y_axis_a, Z_axis_a], dim=0)
        matrix_T = torch.zeros((12, 12), dtype=torch.float32, device=device)
        for i in range(0, 12, 3):
            matrix_T[i:i + 3, i:i + 3] = lambda_matrix
        return matrix_T

# 组装全局刚度矩阵
def assemble_stiffness_matrix(beams, n_nodes, n_dof_per_node, connectivity):
    total_dof = n_nodes * n_dof_per_node
    K_global = torch.zeros((total_dof, total_dof), dtype=torch.float32, device=device)
    
    for idx, (i, j) in enumerate(connectivity):
        Matrix_T = beams[idx].System_Transform()
        K_element = torch.matmul(torch.transpose(Matrix_T, 0, 1),
                                 torch.matmul(beams[idx].get_element_stiffness_matrix(), Matrix_T))

        start_idx = (i - 1) * n_dof_per_node
        end_idx = (j - 1) * n_dof_per_node
        K_global[start_idx:start_idx + 6, start_idx:start_idx + 6] += K_element[0:6, 0:6]
        K_global[end_idx:end_idx + 6, end_idx:end_idx + 6] += K_element[6:12, 6:12]
        K_global[start_idx:start_idx + 6, end_idx:end_idx + 6] += K_element[0:6, 6:12]
        K_global[end_idx:end_idx + 6, start_idx:start_idx + 6] += K_element[6:12, 0:6]

    return K_global

# 鲁棒求解器
def robust_solve(K_global, F, fixed_dof, max_attempts=3):
    attempts = 0
    while attempts < max_attempts:
        reg = 1e-6 * torch.eye(K_global.shape[0], device=K_global.device)
        reg[fixed_dof, fixed_dof] = 0
        K_reg = K_global + reg
        
        try:
            displacements = torch.linalg.solve(
                K_reg.to(torch.float64), 
                F.to(torch.float64)
            )
            return displacements.to(K_global.dtype)
        except RuntimeError:
            diag = torch.diag(K_global)
            extreme_mask = (diag > 1e12) & (~torch.isin(torch.arange(len(diag)), torch.tensor(fixed_dof)))  
            K_reg[extreme_mask] = 0
            K_reg[:, extreme_mask] = 0
            K_reg[extreme_mask, extreme_mask] = 1e12
            
            K_reg[fixed_dof, :] = 0
            K_reg[:, fixed_dof] = 0
            K_reg[fixed_dof, fixed_dof] = 1e10
            
            try:
                displacements, info = torch.linalg.cg(
                    K_reg.to(torch.float64),
                    F.to(torch.float64),
                    maxiter=5000,
                    atol=1e-6
                )
                if info > 0:
                    raise RuntimeError("CG failed")
                return displacements.to(K_global.dtype)
            except:
                K_pinv = torch.linalg.pinv(K_reg)
                K_pinv[fixed_dof, :] = 0
                displacements = K_pinv @ F
                print("Warning: Used pseudo-inverse, accuracy may be reduced")
                return displacements
        attempts += 1
    
    raise RuntimeError("Failed to solve linear system")

# 应变能计算
def calculate_strain_energy(node_coords, connectivity, fixed_dof, F):
    Beam_lens = []
    beams = []
    for connection in connectivity:
        node_1_coords = node_coords[connection[0] - 1]
        node_2_coords = node_coords[connection[1] - 1]
        beam = Beam(node_coordinates=torch.stack([node_1_coords, node_2_coords]),
                    R=D_radius, young_modulus=D_young_modulus,
                    shear_modulus=D_shear_modulus, poisson_ratio=D_poisson_ratio, 
                    Beta_a=cross_section_angle_a, Beta_b=cross_section_angle_b)
        beams.append(beam)
        Beam_lens.append(beam.length)
    
    K_global = assemble_stiffness_matrix(beams, n_nodes=len(node_coords), 
                                        n_dof_per_node=6, connectivity=connectivity)
    K_global[fixed_dof, :] = 0
    K_global[:, fixed_dof] = 0
    K_global[fixed_dof, fixed_dof] = 1e10

    displacements = robust_solve(K_global, F, fixed_dof)

    strain_energy_list = []
    ASE_list = []
    Local_d = torch.zeros(len(connectivity), 12, dtype=torch.float32, device=device)
    
    for n, (i, j) in enumerate(connectivity):
        matrix_T = beams[n].System_Transform()
        Tep_displacements = torch.cat(
            [displacements[6 * (i - 1):6 * (i - 1) + 6], 
             displacements[6 * (j - 1):6 * (j - 1) + 6]], dim=0)
        Local_d_n = torch.matmul(Tep_displacements, matrix_T.T)
        Local_d[n, :] = Local_d_n.clone()
        K_l = beams[n].get_element_stiffness_matrix()
        
        # 总应变能
        strain_energy = 0.5 * torch.matmul(Local_d_n, torch.matmul(K_l, Local_d_n.reshape(-1, 1)))
        strain_energy_list.append(strain_energy)
        
        # 轴向应变能 (ASE)
        axial_energy = 0.5 * (Local_d_n[0]-Local_d_n[6]) * beams[n].S_u * (Local_d_n[0]-Local_d_n[6])
        ASE_list.append(axial_energy)
    
    Strain_energy = torch.stack(strain_energy_list)
    ASE = torch.stack(ASE_list)
    Ratio = torch.sum(ASE) / torch.sum(Strain_energy)
    
    return Ratio.item()

# 主函数：计算并绘制Ratio分布
def main():
    # 采样设置
    resolution = 30  # 网格分辨率
    z2_values = np.linspace(0, 2, resolution)
    z3_values = np.linspace(0, 2, resolution)
    ratio_grid = np.zeros((resolution, resolution))
    
    # 固定节点
    fixed_nodes = [0, 3]  # 节点1和节点4（索引从0开始）
    
    # 连接关系
    connectivity = [(1, 2), (2, 3), (3, 4)]  # 杆件连接
    
    # 荷载设置（在节点2和3施加垂直荷载）
    F = torch.zeros(24, dtype=torch.float32, device=device)  # 4节点 * 6自由度
    F[6*1 + 2] = -1000  # 节点2的z方向力 (索引: 1*6 + 2 = 8)
    F[6*2 + 2] = -1000  # 节点3的z方向力 (索引: 2*6 + 2 = 14)
    
    # 固定自由度（节点1和节点4的所有自由度）
    fixed_dof = []
    for node_idx in fixed_nodes:
        fixed_dof.extend(range(node_idx*6, node_idx*6+6))
    
    # 进度计数器
    total_points = resolution * resolution
    completed = 0
    
    # 遍历所有采样点
    for i, z2 in enumerate(z2_values):
        for j, z3 in enumerate(z3_values):
            # 节点坐标 (x, y=0, z)
            node_coords = torch.tensor([
                [0.0, 0.0, 0.0],   # 节点1
                [2.0, 0.0, z2],    # 节点2
                [4.0, 0.0, z3],    # 节点3
                [6.0, 0.0, 0.0]    # 节点4
            ], dtype=torch.float32, device=device)
            
            try:
                ratio = calculate_strain_energy(node_coords, connectivity, fixed_dof, F)
                ratio_grid[j, i] = ratio  # j对应y轴(z3)，i对应x轴(z2)
            except Exception as e:
                print(f"Error at (z2={z2:.2f}, z3={z3:.2f}): {str(e)}")
                ratio_grid[j, i] = np.nan
            
            completed += 1
            if completed % 50 == 0:
                print(f"Progress: {completed}/{total_points} ({completed/total_points*100:.1f}%)")
    
    # 绘制结果
    plt.figure(figsize=(12, 10))
    
    # 创建自定义颜色映射
    colors = [(0.8, 0.0, 0.0), (1.0, 0.8, 0.0), (0.0, 0.6, 0.0)]  # 红 -> 黄 -> 绿
    cmap = LinearSegmentedColormap.from_list("custom", colors, N=256)
    
    # 绘制应变能比率热图
    im = plt.imshow(ratio_grid, extent=[0, 2, 0, 2], origin='lower', 
                   cmap=cmap, vmin=0, vmax=1)
    plt.colorbar(im, label='Axial Strain Energy Ratio (ASE/Total)')
    
    # 标记高比率区域 (Ratio > 0.8)
    high_ratio_mask = ratio_grid > 0.8
    z2_mesh, z3_mesh = np.meshgrid(z2_values, z3_values)
    plt.contour(z2_mesh, z3_mesh, high_ratio_mask.astype(float), levels=[0.5], 
                colors='blue', linewidths=2)
    
    # 添加标记点
    plt.plot(0.0, 0.0, 'kx', markersize=8, label='Flat Structure (0.0, 0.0)')
    
    # 设置标题和标签
    plt.title('Axial Strain Energy Ratio Distribution\n(z2 and z3 in [0, 2] range)', fontsize=14)
    plt.xlabel('Node 2 z-coordinate (z2)', fontsize=12)
    plt.ylabel('Node 3 z-coordinate (z3)', fontsize=12)
    
    # 添加网格和图例
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.legend(loc='upper left')
    
    # 添加说明文本框
    props = dict(boxstyle='round', facecolor='white', alpha=0.8)
    textstr = '\n'.join((
        'Structure Parameters:',
        f'- Young\'s modulus: {D_young_modulus/1e9:.1f} GPa',
        f'- Radius: {D_radius:.2f} m',
        f'- Load: 1000N downward at nodes 2 & 3',
        f'- Fixed nodes: 1 & 4'))
    plt.text(1.05, 0.05, textstr, transform=plt.gca().transAxes, 
             fontsize=10, verticalalignment='bottom', bbox=props)
    
    plt.tight_layout()
    plt.savefig('strain_energy_ratio_distribution.png', dpi=300)
    plt.show()
    
    # 打印高比率区域统计数据
    high_ratio_points = np.sum(high_ratio_mask)
    total_points = resolution * resolution
    print(f"\nResults Summary:")
    print(f"Total points: {total_points}")
    print(f"Points with Ratio > 0.8: {high_ratio_points} ({high_ratio_points/total_points*100:.1f}%)")
    print(f"Minimum Ratio: {np.nanmin(ratio_grid):.4f}")
    print(f"Maximum Ratio: {np.nanmax(ratio_grid):.4f}")
    print(f"Average Ratio: {np.nanmean(ratio_grid):.4f}")

if __name__ == "__main__":
    main()
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

# 设置设备
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# 常量定义
D_radius = 0.75
D_young_modulus = 10e9 
D_shear_modulus = 0.7e9 
D_poisson_ratio = 0.3
cross_section_angle_a = 0  
cross_section_angle_b = 0  
a_small_number = 1e-10

# 旋转函数
def rotation(v, k, theta):
    k = k / torch.norm(k)  # 归一化
    cross_product = torch.cross(k, v)
    dot_product = torch.dot(k, v)
    theta = torch.tensor(theta, dtype=torch.float32, device=device) if not isinstance(theta, torch.Tensor) else theta
    v_rotated = v * torch.cos(theta) + cross_product * torch.sin(theta) + k * dot_product * (1 - torch.cos(theta))
    return v_rotated

# 梁单元类
class Beam:
    def __init__(self, node_coordinates, R=D_radius, young_modulus=D_young_modulus,
                 shear_modulus=D_shear_modulus, poisson_ratio=D_poisson_ratio, Beta_a=cross_section_angle_a,
                 Beta_b=cross_section_angle_b):
        self.node_coordinates = node_coordinates

        # 材料和几何属性
        self.radius = R
        self.young_modulus = young_modulus
        self.shear_modulus = shear_modulus
        self.poisson_ratio = poisson_ratio

        # 截面属性
        self.length = torch.norm(self.node_coordinates[1] - self.node_coordinates[0])
        self.Iy = (torch.pi * self.radius ** 4) / 4 
        self.Iz = self.Iy
        self.A = torch.pi * self.radius ** 2
        self.J = (torch.pi * self.radius ** 4) / 2

        # 刚度分量
        self.S_u = self.young_modulus * self.A / self.length
        self.S_v1a = 12 * self.young_modulus * self.Iy / (self.length ** 3)
        self.S_v1b = 6 * self.young_modulus * self.Iy / (self.length ** 2)
        self.S_v2a = 12 * self.young_modulus * self.Iz / (self.length ** 3)
        self.S_v2b = 6 * self.young_modulus * self.Iz / (self.length ** 2)
        self.S_theta1a = 6 * self.young_modulus * self.Iy / (self.length ** 2)
        self.S_theta1b = 4 * self.young_modulus * self.Iy / self.length
        self.S_theta1c = 2 * self.young_modulus * self.Iy / self.length
        self.S_theta2a = 6 * self.young_modulus * self.Iz / (self.length ** 2)
        self.S_theta2b = 4 * self.young_modulus * self.Iz / self.length
        self.S_theta2c = 2 * self.young_modulus * self.Iz / self.length
        self.S_Tr = self.shear_modulus * self.J / self.length

        # 截面旋转
        self.Beta_a = Beta_a
        self.Beta_b = Beta_b

    def get_element_stiffness_matrix(self):
        K_element = torch.tensor([
            [self.S_u, 0, 0, 0, 0, 0, -self.S_u, 0, 0, 0, 0, 0],
            [0, self.S_v1a, 0, 0, 0, self.S_theta1a, 0, -self.S_v1a, 0, 0, 0, self.S_theta1a],
            [0, 0, self.S_v2a, 0, -self.S_theta2a, 0, 0, 0, -self.S_v2a, 0, -self.S_theta2a, 0],
            [0, 0, 0, self.S_Tr, 0, 0, 0, 0, 0, -self.S_Tr, 0, 0],
            [0, 0, -self.S_v2b, 0, self.S_theta2b, 0, 0, 0, self.S_v2b, 0, self.S_theta2c, 0],
            [0, self.S_v1b, 0, 0, 0, self.S_theta1b, 0, -self.S_v1b, 0, 0, 0, self.S_theta1c],
            [-self.S_u, 0, 0, 0, 0, 0, self.S_u, 0, 0, 0, 0, 0],
            [0, -self.S_v1a, 0, 0, 0, -self.S_theta1a, 0, self.S_v1a, 0, 0, 0, -self.S_theta1a],
            [0, 0, -self.S_v2a, 0, self.S_theta2a, 0, 0, 0, self.S_v2a, 0, self.S_theta2a, 0],
            [0, 0, 0, -self.S_Tr, 0, 0, 0, 0, 0, self.S_Tr, 0, 0],
            [0, 0, -self.S_v2b, 0, self.S_theta2c, 0, 0, 0, self.S_v2b, 0, self.S_theta2b, 0],
            [0, self.S_v1b, 0, 0, 0, self.S_theta1c, 0, -self.S_v1b, 0, 0, 0, self.S_theta1b],
        ], dtype=torch.float32, device=device)

        return K_element

    def System_Transform(self):
        vector_x = self.node_coordinates[1, 0] - self.node_coordinates[0, 0]
        vector_y = self.node_coordinates[1, 1] - self.node_coordinates[0, 1]
        vector_z = self.node_coordinates[1, 2] - self.node_coordinates[0, 2]
        length = torch.norm(self.node_coordinates[1] - self.node_coordinates[0])
        
        z_value = torch.clamp(vector_z / length, min=-1 + 1e-6, max=1 - 1e-6)
        ceta = torch.acos(z_value)
        value = vector_x / torch.sqrt(vector_y ** 2 + vector_x ** 2 + a_small_number)
        value = torch.clamp(value, min=-1 + 1e-6, max=1 - 1e-6)
        alpha = torch.acos(value)

        Projection_Z_x = - vector_z / length * torch.sin(alpha)
        Projection_Z_y = - vector_z / length * torch.cos(alpha)
        Projection_Z_z = torch.cos(torch.pi / 2 - ceta)

        V_projection = torch.stack([Projection_Z_x, Projection_Z_y, Projection_Z_z])
        X_axis = torch.stack([vector_x / length, vector_y / length, vector_z / length])
        Z_axis_a = rotation(V_projection, X_axis, self.Beta_a)
        Y_axis_a = rotation(Z_axis_a, X_axis, -torch.pi / 2)
        Z_axis_a = Z_axis_a / torch.norm(Z_axis_a)
        Y_axis_a = Y_axis_a / torch.norm(Y_axis_a)

        lambda_matrix = torch.stack([X_axis, Y_axis_a, Z_axis_a], dim=0)
        matrix_T = torch.zeros((12, 12), dtype=torch.float32, device=device)
        for i in range(0, 12, 3):
            matrix_T[i:i + 3, i:i + 3] = lambda_matrix
        return matrix_T

# 组装全局刚度矩阵
def assemble_stiffness_matrix(beams, n_nodes, n_dof_per_node, connectivity):
    total_dof = n_nodes * n_dof_per_node
    K_global = torch.zeros((total_dof, total_dof), dtype=torch.float32, device=device)
    
    for idx, (i, j) in enumerate(connectivity):
        Matrix_T = beams[idx].System_Transform()
        K_element = torch.matmul(torch.transpose(Matrix_T, 0, 1),
                                 torch.matmul(beams[idx].get_element_stiffness_matrix(), Matrix_T))

        start_idx = (i - 1) * n_dof_per_node
        end_idx = (j - 1) * n_dof_per_node
        K_global[start_idx:start_idx + 6, start_idx:start_idx + 6] += K_element[0:6, 0:6]
        K_global[end_idx:end_idx + 6, end_idx:end_idx + 6] += K_element[6:12, 6:12]
        K_global[start_idx:start_idx + 6, end_idx:end_idx + 6] += K_element[0:6, 6:12]
        K_global[end_idx:end_idx + 6, start_idx:start_idx + 6] += K_element[6:12, 0:6]

    return K_global

# 鲁棒求解器
def robust_solve(K_global, F, fixed_dof, max_attempts=3):
    attempts = 0
    while attempts < max_attempts:
        reg = 1e-6 * torch.eye(K_global.shape[0], device=K_global.device)
        reg[fixed_dof, fixed_dof] = 0
        K_reg = K_global + reg
        
        try:
            displacements = torch.linalg.solve(
                K_reg.to(torch.float64), 
                F.to(torch.float64)
            )
            return displacements.to(K_global.dtype)
        except RuntimeError:
            diag = torch.diag(K_global)
            extreme_mask = (diag > 1e12) & (~torch.isin(torch.arange(len(diag)), torch.tensor(fixed_dof)))  
            K_reg[extreme_mask] = 0
            K_reg[:, extreme_mask] = 0
            K_reg[extreme_mask, extreme_mask] = 1e12
            
            K_reg[fixed_dof, :] = 0
            K_reg[:, fixed_dof] = 0
            K_reg[fixed_dof, fixed_dof] = 1e10
            
            try:
                displacements, info = torch.linalg.cg(
                    K_reg.to(torch.float64),
                    F.to(torch.float64),
                    maxiter=5000,
                    atol=1e-6
                )
                if info > 0:
                    raise RuntimeError("CG failed")
                return displacements.to(K_global.dtype)
            except:
                K_pinv = torch.linalg.pinv(K_reg)
                K_pinv[fixed_dof, :] = 0
                displacements = K_pinv @ F
                print("Warning: Used pseudo-inverse, accuracy may be reduced")
                return displacements
        attempts += 1
    
    raise RuntimeError("Failed to solve linear system")

# 应变能计算
def calculate_strain_energy(node_coords, connectivity, fixed_dof, F):
    Beam_lens = []
    beams = []
    for connection in connectivity:
        node_1_coords = node_coords[connection[0] - 1]
        node_2_coords = node_coords[connection[1] - 1]
        beam = Beam(node_coordinates=torch.stack([node_1_coords, node_2_coords]),
                    R=D_radius, young_modulus=D_young_modulus,
                    shear_modulus=D_shear_modulus, poisson_ratio=D_poisson_ratio, 
                    Beta_a=cross_section_angle_a, Beta_b=cross_section_angle_b)
        beams.append(beam)
        Beam_lens.append(beam.length)
    
    K_global = assemble_stiffness_matrix(beams, n_nodes=len(node_coords), 
                                        n_dof_per_node=6, connectivity=connectivity)
    K_global[fixed_dof, :] = 0
    K_global[:, fixed_dof] = 0
    K_global[fixed_dof, fixed_dof] = 1e10

    displacements = robust_solve(K_global, F, fixed_dof)

    strain_energy_list = []
    ASE_list = []
    Local_d = torch.zeros(len(connectivity), 12, dtype=torch.float32, device=device)
    
    for n, (i, j) in enumerate(connectivity):
        matrix_T = beams[n].System_Transform()
        Tep_displacements = torch.cat(
            [displacements[6 * (i - 1):6 * (i - 1) + 6], 
             displacements[6 * (j - 1):6 * (j - 1) + 6]], dim=0)
        Local_d_n = torch.matmul(Tep_displacements, matrix_T.T)
        Local_d[n, :] = Local_d_n.clone()
        K_l = beams[n].get_element_stiffness_matrix()
        
        # 总应变能
        strain_energy = 0.5 * torch.matmul(Local_d_n, torch.matmul(K_l, Local_d_n.reshape(-1, 1)))
        strain_energy_list.append(strain_energy)
        
        # 轴向应变能 (ASE)
        axial_energy = 0.5 * (Local_d_n[0]-Local_d_n[6]) * beams[n].S_u * (Local_d_n[0]-Local_d_n[6])
        ASE_list.append(axial_energy)
    
    Strain_energy = torch.stack(strain_energy_list)
    ASE = torch.stack(ASE_list)
    Ratio = torch.sum(ASE) / torch.sum(Strain_energy)
    
    return Ratio.item()

# 主函数：计算并绘制Ratio分布
def main():
    # 采样设置
    resolution = 30  # 网格分辨率
    z2_values = np.linspace(0, 2, resolution)
    z3_values = np.linspace(0, 2, resolution)
    ratio_grid = np.zeros((resolution, resolution))
    
    # 固定节点
    fixed_nodes = [0, 3]  # 节点1和节点4（索引从0开始）
    
    # 连接关系
    connectivity = [(1, 2), (2, 3), (3, 4)]  # 杆件连接
    
    # 荷载设置（在节点2和3施加垂直荷载）
    F = torch.zeros(24, dtype=torch.float32, device=device)  # 4节点 * 6自由度
    F[6*1 + 2] = -1000  # 节点2的z方向力 (索引: 1*6 + 2 = 8)
    F[6*2 + 2] = -1000  # 节点3的z方向力 (索引: 2*6 + 2 = 14)
    
    # 固定自由度（节点1和节点4的所有自由度）
    fixed_dof = []
    for node_idx in fixed_nodes:
        fixed_dof.extend(range(node_idx*6, node_idx*6+6))
    
    # 进度计数器
    total_points = resolution * resolution
    completed = 0
    
    # 遍历所有采样点
    for i, z2 in enumerate(z2_values):
        for j, z3 in enumerate(z3_values):
            # 节点坐标 (x, y=0, z)
            node_coords = torch.tensor([
                [0.0, 0.0, 0.0],   # 节点1
                [2.0, 0.0, z2],    # 节点2
                [4.0, 0.0, z3],    # 节点3
                [6.0, 0.0, 0.0]    # 节点4
            ], dtype=torch.float32, device=device)
            
            try:
                ratio = calculate_strain_energy(node_coords, connectivity, fixed_dof, F)
                ratio_grid[j, i] = ratio  # j对应y轴(z3)，i对应x轴(z2)
            except Exception as e:
                print(f"Error at (z2={z2:.2f}, z3={z3:.2f}): {str(e)}")
                ratio_grid[j, i] = np.nan
            
            completed += 1
            if completed % 50 == 0:
                print(f"Progress: {completed}/{total_points} ({completed/total_points*100:.1f}%)")
    
    # 绘制结果
    plt.figure(figsize=(12, 10))
    
    # 创建自定义颜色映射
    colors = [(0.8, 0.0, 0.0), (1.0, 0.8, 0.0), (0.0, 0.6, 0.0)]  # 红 -> 黄 -> 绿
    cmap = LinearSegmentedColormap.from_list("custom", colors, N=256)
    
    # 绘制应变能比率热图
    im = plt.imshow(ratio_grid, extent=[0, 2, 0, 2], origin='lower', 
                   cmap=cmap, vmin=0, vmax=1)
    plt.colorbar(im, label='Axial Strain Energy Ratio (ASE/Total)')
    
    # 标记高比率区域 (Ratio > 0.8)
    high_ratio_mask = ratio_grid > 0.8
    z2_mesh, z3_mesh = np.meshgrid(z2_values, z3_values)
    plt.contour(z2_mesh, z3_mesh, high_ratio_mask.astype(float), levels=[0.5], 
                colors='blue', linewidths=2)
    
    # 添加标记点
    plt.plot(0.0, 0.0, 'kx', markersize=8, label='Flat Structure (0.0, 0.0)')
    
    # 设置标题和标签
    plt.title('Axial Strain Energy Ratio Distribution\n(z2 and z3 in [0, 2] range)', fontsize=14)
    plt.xlabel('Node 2 z-coordinate (z2)', fontsize=12)
    plt.ylabel('Node 3 z-coordinate (z3)', fontsize=12)
    
    # 添加网格和图例
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.legend(loc='upper left')
    
    # 添加说明文本框
    print("\nStructure Parameters:")
    print(f"- Young's modulus: {D_young_modulus/1e9:.1f} GPa")
    print(f"- Radius: {D_radius:.2f} m")
    print(f"- Load: 1000N downward at nodes 2 & 3")
    print(f"- Fixed nodes: 1 & 4")
    
    plt.tight_layout()
    plt.savefig('strain_energy_ratio_distribution.png', dpi=300)
    plt.show()
        
    # 打印高比率区域统计数据
    high_ratio_points = np.sum(high_ratio_mask)
    total_points = resolution * resolution
    print(f"\nResults Summary:")
    print(f"Total points: {total_points}")
    print(f"Points with Ratio > 0.8: {high_ratio_points} ({high_ratio_points/total_points*100:.1f}%)")
    print(f"Minimum Ratio: {np.nanmin(ratio_grid):.4f}")
    print(f"Maximum Ratio: {np.nanmax(ratio_grid):.4f}")
    print(f"Average Ratio: {np.nanmean(ratio_grid):.4f}")

if __name__ == "__main__":
    main()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import LinearSegmentedColormap

# 参数设置
mu_g = 1.0  # 线密度 × 重力加速度
T0 = 2.0    # 顶点水平张力 (T = 2/y 的最低点张力)

# 悬链线方程 (顶点在x=0, y=0)
def catenary(x, a):
    return a * np.cosh(x / a) - a  # 使顶点y(0)=0

# 生成网格
y2_values = np.linspace(0, 2, 50)
y3_values = np.linspace(0, 2, 50)
Y2, Y3 = np.meshgrid(y2_values, y3_values)

# 计算可行区域
feasible = np.zeros_like(Y2)
for i in range(len(y2_values)):
    for j in range(len(y3_values)):
        y2 = y2_values[i]
        y3 = y3_values[j]
        
        if y2 > 0:
            # 牛顿法求解参数a
            a = 1.0  # 初始猜测
            for _ in range(50):  # 增加迭代次数保证收敛
                f = catenary(2, a) - y2
                df = (catenary(2, a + 1e-6) - catenary(2, a)) / 1e-6
                a -= f / df
                if abs(f) < 1e-6:
                    break
            # 检查y3是否匹配
            if abs(catenary(4, a) - y3) < 0.05:
                feasible[j, i] = 1

# 可视化设置
plt.figure(figsize=(10, 8))
ax = plt.gca()

# 绘制可行域
cmap = LinearSegmentedColormap.from_list("feasible", ["white", "darkgreen"], N=2)
im = ax.imshow(feasible, extent=[0, 2, 0, 2], origin='lower', 
               cmap=cmap, alpha=0.7, aspect='auto')

# 添加理论曲线
a_vals = np.linspace(0.1, 10, 100)
y2_curve = catenary(2, a_vals)
y3_curve = catenary(4, a_vals)
ax.plot(y2_curve, y3_curve, 'r-', lw=3, label="Exact Catenary Constraint")

# 标注和美化
ax.set_xlabel("y₂ (Vertical Displacement at x=2)", fontsize=12)
ax.set_ylabel("y₃ (Vertical Displacement at x=4)", fontsize=12)
ax.set_title("Feasible Region for Inverted Catenary Arch\n(Tension Constraint: T₀ = 2)", 
             fontsize=14, pad=20)

# 确保坐标轴显示
ax.spines['bottom'].set_visible(True)
ax.spines['left'].set_visible(True)
ax.spines['bottom'].set_color('black')
ax.spines['left'].set_color('black')
ax.tick_params(axis='both', which='both', length=6, width=1)

# 添加关键点和注释
ax.plot(0, 0, 'ko', ms=8, label="Flat Configuration (a→∞)")
ax.plot(catenary(2, 1.0), catenary(4, 1.0), 'go', ms=8, label="a=1.0 Example")

# 物理方程标注
eq_text = (r"$\frac{T_0}{\mu g} = a = 2$ (Fixed)" + "\n" +
           r"$y(x) = a \left(\cosh\left(\frac{x}{a}\right) - 1\right)$")
ax.text(1.5, 0.1, eq_text, fontsize=11, 
        bbox=dict(facecolor='white', alpha=0.8, edgecolor='gray'))

ax.grid(True, linestyle=':', alpha=0.5)
ax.legend(loc='upper left', framealpha=1)
plt.tight_layout()
plt.show()
```

# 重新formulate 这个问题；把坐标倒置y 坐标0,-2,-2,0
然后这一次把design space 的 parameter 换成 节点2 的x 和 y 坐标；并且sample (60 * 30) 对应 x, y的范围[0, 4] 和 [0.2]; 然后 设置重度 沿轴向方向为 1； 设置力矩为零平衡约束：
举例：
当 x=2,y 等于 -2: element 1 的重力为 根号8 (跟号x方加y方)；
element 1 [node 1, node 2] 重力为 根号8； 作用点在 x = 1 ；y = -1 的 位置
然后做静力分析：
利用这些条件计算作用在节点1 和 节点 4 的支座反力, 举例设置为 fx1 和 fx2
fx1 + fx2 = 跟号8加2加（根号8）-> 最后一项值不变化因为几何固定;
针对节点2 的力矩平衡：
fx1 * 2 - fe1 *1 + fe2 * 1 + fe3 * 3 - fx2*4 （fe分别是三个杆件的重力作用在其中心点； fe3 等于根号8 因为几何不变;
然后你就可以根据fx1 和 fe1 以及 x,y 的左边关系计算 T0了；
也就是tan = y/x = T0/fx1 ;
从而计算出来T0 -> 然后将现在sample 的 x, y 代入catenary公式验证


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve

# 参数设置
mu_g = 1.0  # 重度沿轴向分布
L_total = 6.0  # 总跨度 (x1=0, x4=6)
fixed_y = [0, 0]  # 节点1和4的y坐标 (y1=y4=0)

# 采样设计空间 (节点2的x和y坐标)
x2_samples = np.linspace(0, 4, 60)  # 避免x2=0导致除零错误
y2_samples = np.linspace(0.1, 2, 30)  # y2>0 (实际物理y向下为正)
X2, Y2 = np.meshgrid(x2_samples, y2_samples)

# 存储验证结果
feasible = np.zeros_like(X2, dtype=bool)

for i in range(len(x2_samples)):
    for j in range(len(y2_samples)):
        x2 = x2_samples[i]
        y2 = -y2_samples[j]  # 转换为物理坐标 (y向下为正)
        
        # 几何参数
        x_coords = [0, x2, 4, 6]  # 节点x坐标
        y_coords = [0, y2, y2, 0]  # 节点y坐标 (对称)
        
        # 计算各杆件长度和重力 (fe = mu_g * length)
        lengths = [
            np.sqrt((x2-0)**2 + (y2-0)**2),    # 杆件1 (节点1-2)
            np.sqrt((4-x2)**2 + (y2-y2)**2),    # 杆件2 (节点2-3)
            np.sqrt((6-4)**2 + (0-y2)**2)       # 杆件3 (节点3-4)
        ]
        fe1, fe2, fe3 = mu_g * np.array(lengths)
        
        # 静力平衡方程求解 fx1, fx2 (水平支座反力)
        # 总力平衡: fx1 + fx2 = fe1 + fe2 + fe3
        # 力矩平衡 (绕节点1): fx2*6 = fe1*(x2/2) + fe2*(x2 + (4-x2)/2) + fe3*(4 + 1)
        A = np.array([[1, 1], [0, 6]])
        b = np.array([fe1 + fe2 + fe3, 
                     fe1*x2/2 + fe2*(x2 + (4-x2)/2) + fe3*5])
        fx1, fx2 = np.linalg.solve(A, b)
        
        # 计算T0 (杆件1的轴向力水平分量)
        theta = np.arctan2(abs(y2), x2)  # 杆件1与水平夹角
        T0 = fx1 / np.cos(theta)  # T0 = fx1 / cosθ
        
        # 验证悬链线约束 (a = T0 / (mu_g))
        a = T0 / mu_g
        # 检查节点3的y坐标是否符合悬链线方程
        y3_catenary = a * (np.cosh(4/a) - np.cosh((4-x2)/a)) + y2
        if abs(y3_catenary - y2) < 0.1:  # 允许误差
            feasible[j, i] = True

# 可视化
plt.figure(figsize=(12, 6))
plt.contourf(X2, Y2, feasible, levels=[-0.5, 0.5, 1.5], colors=['white', 'lime'], alpha=0.6)
plt.scatter(X2[feasible], Y2[feasible], c='red', s=10, label="Feasible Designs")
plt.xlabel("Node 2 x-coordinate (x₂ ∈ [0,4])", fontsize=12)
plt.ylabel("Node 2 |y-coordinate| (|y₂| ∈ [0,2], ↓ positive)", fontsize=12)
plt.title("Feasible Region for Inverted Catenary Arch\n(Valid designs satisfy static equilibrium + catenary constraint)", fontsize=14)

# 添加示例杆件示意图
ax_inset = plt.axes([0.7, 0.7, 0.2, 0.2])
sample_idx = np.argmax(feasible)  # 取一个可行解
i, j = np.unravel_index(sample_idx, feasible.shape)
x2 = X2[i, j]
y2 = -Y2[i, j]
x_nodes = [0, x2, 4, 6]
y_nodes = [0, y2, y2, 0]
ax_inset.plot(x_nodes, y_nodes, 'bo-')
ax_inset.set_title(f"Example: x₂={x2:.1f}, y₂={-y2:.1f}")
ax_inset.set_xlim(-1, 7)
ax_inset.set_ylim(-3, 1)
ax_inset.set_aspect('equal')

plt.grid(True, linestyle=':', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

# 参数设置
mu_g = 1.0  # 重度沿轴向分布
L_total = 6.0  # 总跨度 (x1=0, x4=6)
fixed_y = [0, 0]  # 节点1和4的y坐标 (y1=y4=0)

# 采样设计空间 (节点2的x和y坐标)
x2_samples = np.linspace(0.1, 4, 60)  # 避免x2=0导致除零错误
y2_samples = np.linspace(0.1, 2, 30)  # y2>0 (实际物理y向下为正)

X2, Y2 = np.meshgrid(x2_samples, y2_samples)

# 存储验证结果
feasible = np.zeros_like(X2, dtype=bool)

for i in range(len(x2_samples)):
    for j in range(len(y2_samples)):
        x2 = x2_samples[i]
        y2 = -y2_samples[j]  # 转换为物理坐标 (y向下为正)
        
        # 几何参数
        x_coords = [0, x2, 4, 6]
        y_coords = [0, y2, y2, 0]
        
        # 计算各杆件长度和重力 (fe = mu_g * length)
        lengths = [
            np.sqrt((x2-0)**2 + (y2-0)**2),    # 杆件1 (节点1-2)
            np.sqrt((4-x2)**2 + (y2-y2)**2),    # 杆件2 (节点2-3)
            np.sqrt((6-4)**2 + (0-y2)**2)       # 杆件3 (节点3-4)
        ]
        fe1, fe2, fe3 = mu_g * np.array(lengths)
        
        # 静力平衡方程求解 Fy1 (支座1的竖向反力)
        Fy1 = fe1 + fe2 + fe3  # 总竖向力平衡
        
        # 计算杆件1的轴向力 T0 (修正后)
        L1 = lengths[0]
        theta = np.arctan2(abs(y2), x2)
        T0 = (Fy1 * L1) / abs(y2)  # T0 = Fy1 / sinθ
        
        # 验证悬链线约束 (a = T0 / (mu_g))
        a = T0 / mu_g
        y3_catenary = a * (np.cosh(4/a) - np.cosh((4-x2)/a)) + y2
        if abs(y3_catenary - y2) < 0.1:  # 允许误差
            feasible[j, i] = True

# 创建自定义颜色映射
colors = [(0.8, 0.0, 0.0), (1.0, 0.8, 0.0), (0.0, 0.6, 0.0)]  # 红 -> 黄 -> 绿
cmap = LinearSegmentedColormap.from_list("custom", colors, N=256)

# 创建新的可视化
plt.figure(figsize=(12, 10))

# 绘制可行区域热图 (将布尔值转换为0/1)
im = plt.imshow(feasible.astype(float), extent=[0.1, 4, 0.1, 2], 
               origin='lower', cmap=cmap, vmin=0, vmax=1, aspect='auto')
plt.colorbar(im, label='Feasibility (1=Feasible, 0=Infeasible)')

# 标记高可行性区域
high_feasible_mask = feasible > 0
x2_mesh, y2_mesh = np.meshgrid(x2_samples, y2_samples)
plt.contour(x2_mesh, y2_mesh, high_feasible_mask.astype(float), levels=[0.5], 
            colors='blue', linewidths=2)

# 添加标记点
plt.plot(2.0, 1.0, 'kx', markersize=8, label='Example Point (2.0, 1.0)')

# 设置标题和标签
plt.title('Feasible Region for Inverted Catenary Arch\n(x₂ ∈ [0.1,4], |y₂| ∈ [0.1,2])', fontsize=14)
plt.xlabel('Node 2 x-coordinate (x₂)', fontsize=12)
plt.ylabel('Node 2 |y-coordinate| (|y₂|, ↓ positive)', fontsize=12)

# 添加网格和图例
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(loc='upper right')

# 添加说明文本框
plt.text(0.5, 1.9, 
         f"Parameters:\n- μ_g = {mu_g}\n- L_total = {L_total}\n- Fixed y1,y4 = {fixed_y}",
         bbox=dict(facecolor='white', alpha=0.8))

# 添加示例结构示意图
ax_inset = plt.axes([0.7, 0.7, 0.2, 0.2])
sample_idx = np.argmax(feasible.sum(axis=1))  # 取一个典型可行解
i, j = sample_idx, np.argmax(feasible[sample_idx])
x2 = X2[i, j]
y2 = -Y2[i, j]
x_nodes = [0, x2, 4, 6]
y_nodes = [0, y2, y2, 0]
ax_inset.plot(x_nodes, y_nodes, 'bo-')
ax_inset.set_title(f"Example: x₂={x2:.1f}, y₂={-y2:.1f}")
ax_inset.set_xlim(-1, 7)
ax_inset.set_ylim(-3, 1)
ax_inset.set_aspect('equal')

plt.tight_layout()
plt.savefig('catenary_feasible_region.png', dpi=300)
plt.show()

# 打印统计信息
feasible_points = np.sum(feasible)
total_points = len(x2_samples) * len(y2_samples)
print(f"\nResults Summary:")
print(f"Total points: {total_points}")
print(f"Feasible points: {feasible_points} ({feasible_points/total_points*100:.1f}%)")
```
