```python
import json
import matplotlib.pyplot as plt
import numpy as np

# File names and labels
files = {
    'FDM_data.json': 'FDM',
    'GRAVITY_data.json': 'Gravity',
    'LC1_data.json': 'LC1',
    'LC2_data.json': 'LC2',
}

# Colors for strain energy components
colors = {
    'strain_energy_g': 'black',
    'strain_energy_l1': 'blue',
    'strain_energy_l2': 'red'
}

all_data = {}
max_iterations = 0

for file, label in files.items():
    with open(file, 'r') as f:
        data = json.load(f)
    
    # Get total number of iterations (epochs)
    total_iterations = data['metadata']['parameters']['epochs']
    
    # Initialize lists to store strain energy values
    strain_energy_g = []
    strain_energy_l1 = []
    strain_energy_l2 = []
    
    # Extract values for each iteration
    for iteration_num in range(total_iterations + 1):  # +1 to include the last iteration
        iteration_data = data['iterations'][iteration_num]  # Get data for this iteration
        strain_energy_g.append(iteration_data['strain_energy_g'])
        strain_energy_l1.append(iteration_data['strain_energy_l1'])
        strain_energy_l2.append(iteration_data['strain_energy_l2'])
    
    # Store the data
    all_data[label] = {
        'strain_energy_g': strain_energy_g,
        'strain_energy_l1': strain_energy_l1,
        'strain_energy_l2': strain_energy_l2
    }
    
    # Update maximum iteration length
    max_iterations = max(max_iterations, total_iterations)

# Plotting
plt.figure(figsize=(12, 8))

# Plot all curves
for label, energies in all_data.items():
    for key, values in energies.items():
        x = np.arange(len(values))  # Create x-axis values
        plt.plot(x, values, 
                color=colors[key],
                linestyle='--' if 'l1' in key else '-',
                linewidth=1.5,
                label=f'{label}_{key}' if label == 'FDM' else None)

# Mark key points (first, every 50th, last)
marker_points = [0] + list(range(50, max_iterations + 1, 50)) + [max_iterations]

for label, energies in all_data.items():
    for key, values in energies.items():
        for point in marker_points:
            if point < len(values):  # Check if point exists in this dataset
                plt.scatter(point, values[point], 
                           color=colors[key],
                           zorder=5)
                if point in [0, max_iterations]:  # Annotate first/last points
                    plt.text(point, values[point],
                            f'{values[point]:.2f}',
                            ha='right' if point == max_iterations else 'left',
                            va='bottom',
                            bbox=dict(facecolor='white', alpha=0.8))

# Customize plot
plt.xlabel('Iterations')
plt.ylabel('Strain Energy')
plt.title('Strain Energy Components Across Optimization Histories')
plt.grid(True, linestyle='--', alpha=0.6)

# Show legend for first file only to avoid clutter
handles, labels = plt.gca().get_legend_handles_labels()
plt.legend(handles[:3], labels[:3], loc='upper right')

plt.tight_layout()
plt.savefig('strain_energy_history.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
import json
import matplotlib.pyplot as plt
import numpy as np

# 文件配置
files = {
    'FDM_data.json': {'label': 'FDM', 'marker': 'o'},
    'GRAVITY_data.json': {'label': 'Gravity', 'marker': 's'}, 
    'LC1_data.json': {'label': 'LC1', 'marker': '^'},
    'LC2_data.json': {'label': 'LC2', 'marker': 'D'}
}

# 应变能分量配置
strain_components = {
    'strain_energy_g': {'color': 'black', 'linestyle': '-', 'label': 'Gravity'},
    'strain_energy_l1': {'color': 'blue', 'linestyle': '--', 'label': 'Load case 1'},
    'strain_energy_l2': {'color': 'red', 'linestyle': ':', 'label': 'Load case 2'}
}

# 数据收集
all_data = {}
max_iterations = 0

plt.figure(figsize=(14, 8))

for file, config in files.items():
    try:
        with open(file) as f:
            data = json.load(f)
        
        total_iter = data['metadata']['parameters']['epochs']
        iterations_data = data['iterations']
        
        # 收集数据
        components = {}
        for comp in strain_components:
            components[comp] = [iterations_data[i][comp] for i in range(total_iter + 1)]
        
        all_data[config['label']] = components
        max_iterations = max(max_iterations, total_iter)
        
        # 绘制曲线
        for comp, style in strain_components.items():
            plt.plot(range(total_iter + 1), 
                    components[comp],
                    color=style['color'],
                    linestyle=style['linestyle'],
                    linewidth=1.5,
                    marker=config['marker'],
                    markevery=int(total_iter/10),  # 每10%标记一个点
                    markersize=6,
                    label=f"{config['label']} {style['label']}")

    except FileNotFoundError:
        print(f"警告: 文件 {file} 未找到，已跳过")
        continue

# 图表装饰
plt.xlabel('迭代次数', fontsize=12)
plt.ylabel('应变能', fontsize=12)
plt.title('不同优化方案的应变能变化对比', fontsize=14, pad=20)
plt.grid(True, linestyle='--', alpha=0.6)

# 智能图例布局
handles, labels = plt.gca().get_legend_handles_labels()
plt.legend(handles, labels, 
           bbox_to_anchor=(1.05, 1), 
           loc='upper left',
           framealpha=0.9)

# 标记关键点
for label, data in all_data.items():
    for comp, values in data.items():
        # 标记起点和终点
        for point in [0, len(values)-1]:
            plt.scatter(point, values[point], 
                       color=strain_components[comp]['color'],
                       s=80, zorder=5)
            plt.text(point, values[point], 
                    f"{values[point]:.1f}",
                    ha='right' if point == len(values)-1 else 'left',
                    va='bottom',
                    bbox=dict(facecolor='white', alpha=0.8))

plt.tight_layout()
plt.savefig('应变能对比分析.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import json
import matplotlib.pyplot as plt
import numpy as np

# 文件配置
files = {
    'FDM_data.json': {'label': 'FDM', 'marker': 'o', 'color': '#1f77b4'},
    'GRAVITY_data.json': {'label': 'Gravity', 'marker': 's', 'color': '#ff7f0e'}, 
    'LC1_data.json': {'label': 'LC1', 'marker': '^', 'color': '#2ca02c'},
    'LC2_data.json': {'label': 'LC2', 'marker': 'D', 'color': '#d62728'}
}

# 应变能分量配置
components = {
    'strain_energy_g': {'name': 'Compliance under gravity', 'unit': 'J'},
    'strain_energy_l1': {'name': 'Compliance under loadcase 1', 'unit': 'J'},
    'strain_energy_l2': {'name': 'Compliance under loadcase 2', 'unit': 'J'}
}

# 数据收集
all_data = {}
max_iterations = 0

for file, config in files.items():
    try:
        with open(file) as f:
            data = json.load(f)
        
        total_iter = data['metadata']['parameters']['epochs']
        iterations_data = data['iterations']
        
        # 收集所有分量数据
        all_data[config['label']] = {
            'strain_energy_g': [iterations_data[i]['strain_energy_g'] for i in range(total_iter + 1)],
            'strain_energy_l1': [iterations_data[i]['strain_energy_l1'] for i in range(total_iter + 1)],
            'strain_energy_l2': [iterations_data[i]['strain_energy_l2'] for i in range(total_iter + 1)]
        }
        max_iterations = max(max_iterations, total_iter)
        
    except FileNotFoundError:
        print(f"警告: 文件 {file} 未找到，已跳过")
        continue

# 创建三个子图
fig, axes = plt.subplots(3, 1, figsize=(12, 18), sharex=True)

# 为每个分量绘制独立图表
for ax_idx, (comp_key, comp_config) in enumerate(components.items()):
    ax = axes[ax_idx]
    
    # 绘制所有数据源曲线
    for label, data in all_data.items():
        config = next(cfg for cfg in files.values() if cfg['label'] == label)
        values = data[comp_key]
        
        ax.plot(range(len(values)), values,
                color=config['color'],
                linestyle='-',
                linewidth=2,
                marker=config['marker'],
                markevery=int(len(values)/10),
                markersize=8,
                label=label)
        
        # 标记起点和终点
        for point in [0, len(values)-1]:
            ax.scatter(point, values[point],
                      color=config['color'],
                      s=100, zorder=5)
            ax.text(point, values[point],
                   f"{values[point]:.1f}",
                   ha='right' if point == len(values)-1 else 'left',
                   va='bottom' if values[point] > np.mean(values) else 'top',
                   bbox=dict(facecolor='white', alpha=0.8))
    
    # 图表装饰
    ax.set_title(comp_config['name'], fontsize=14, pad=10)
    ax.set_ylabel(f"{comp_config['name']} ({comp_config['unit']})", fontsize=12)
    ax.grid(True, linestyle='--', alpha=0.6)
    ax.legend(loc='upper right')
    
    # 仅最下方图表显示x轴标签
    if ax_idx == 2:
        ax.set_xlabel('Iteration Number', fontsize=12)

plt.tight_layout(pad=3.0)
plt.savefig('Strain_Energy_Components.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import json

def modify_epochs_in_file(file_path, new_epochs):
    """
    永久修改JSON文件中的metadata->parameters->epochs参数
    :param file_path: JSON文件路径
    :param new_epochs: 新的epochs值
    """
    try:
        # 读取原始文件
        with open(file_path, 'r') as f:
            data = json.load(f)
        
        # 检查路径是否存在
        if 'metadata' in data and 'parameters' in data['metadata']:
            # 修改epochs参数
            data['metadata']['parameters']['epochs'] = new_epochs
            
            # 保存修改后的文件
            with open(file_path, 'w') as f:
                json.dump(data, f, indent=2)
                
            print(f"成功修改 {file_path} 的epochs为 {new_epochs}")
        else:
            print(f"文件结构不符合预期，未找到metadata->parameters路径")
            
    except FileNotFoundError:
        print(f"错误: 文件 {file_path} 未找到")
    except json.JSONDecodeError:
        print(f"错误: 文件 {file_path} 不是有效的JSON格式")
    except Exception as e:
        print(f"修改 {file_path} 时出错: {str(e)}")

# 使用示例
if __name__ == "__main__":
    # 要修改的文件和新的epochs值
    target_file = "Moo_data.json"
    new_epochs_value = 338 # 设置为你想要的新值
    
    # 执行修改
    modify_epochs_in_file(target_file, new_epochs_value)
```

```python
import json
import matplotlib.pyplot as plt
import numpy as np

# Configure global plot settings with Times New Roman for all text
plt.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Times New Roman'],
    'font.size': 11,
    'axes.titlesize': 14,
    'axes.titleweight': 'bold',
    'axes.labelweight': 'bold',
    'xtick.labelsize': 11,
    'ytick.labelsize': 11,
    'legend.fontsize': 11,
    'axes.grid': False,
})

# File configuration with non-primary colors (marker parameters removed)
files = {
    'FDM_data.json': {'label': 'FDM OPT', 'color': '#4e79a7', 'marker': '1'},
    'FEg_data.json': {'label': 'FE Gravity* OPT', 'color': '#76b7b2', 'marker': '1'},
    'GRAVITY_data.json': {'label': 'Gravity* OPT', 'color': '#e15759', 'marker': '1'},
    'LC1_data.json': {'label': 'LC1* OPT', 'color': '#59a14f', 'marker': '2'},
    'LC2_data.json': {'label': 'LC2* OPT', 'color': '#edc948', 'marker': '3'},
}
# Strain energy components
components = {
    'strain_energy_g': {'name': 'Strain energies Under Gravity', 'unit': 'KJ'},
    'strain_energy_l1': {'name': 'Strain energies Under Load Case 1', 'unit': 'KJ'},
    'strain_energy_l2': {'name': 'Strain energies Under Load Case 2', 'unit': 'KJ'}
}

# Data collection
all_data = {}
max_iterations = 0

for file, config in files.items():
    try:
        with open(file) as f:
            data = json.load(f)
        
        total_iter = data['metadata']['parameters']['epochs']
        iterations_data = data['iterations']
        
        all_data[config['label']] = {
            'strain_energy_g': [iterations_data[i]['strain_energy_g'] for i in range(total_iter + 1)],
            'strain_energy_l1': [iterations_data[i]['strain_energy_l1'] for i in range(total_iter + 1)],
            'strain_energy_l2': [iterations_data[i]['strain_energy_l2'] for i in range(total_iter + 1)]
        }
        max_iterations = max(max_iterations, total_iter)
        
    except FileNotFoundError:
        print(f"Warning: File {file} not found, skipped")
        continue

# Create 2x2 grid of subplots
fig, axes = plt.subplots(2, 2, figsize=(12, 12))
fig.subplots_adjust(hspace=0.3, wspace=0.3)
axes = axes.ravel()
siz = 20
# Subplot labels
subplot_labels = ['(1)', '(2)', '(3)', '(4)']

# Plot first three components with normalized values
for ax_idx, (comp_key, comp_config) in enumerate(components.items()):
    ax = axes[ax_idx]
    first_curve = True
    
    for label, data in all_data.items():
        config = next(cfg for cfg in files.values() if cfg['label'] == label)
        values = data[comp_key]
        normalized_values = [v / values[0] for v in values]
        
        # 确定线条样式：当前子图匹配的数字用实线，否则虚线
        subplot_num = ax_idx + 1
        line_style = '-' if str(subplot_num) in config.get('marker', '') else '--'
        
        # 绘制曲线（实际显示虚线/实线）
        ax.plot(
            range(len(values)), normalized_values,
            color=config['color'],
            linestyle=line_style,  # 实际线条样式
            linewidth=2,
            label=label  # 标签用于图例
        )
        
        # 标记起点和终点（保持原代码）
        ax.scatter(0, normalized_values[0], color=config['color'], s=10, zorder=5)
        ax.scatter(len(values)-1, normalized_values[-1], color=config['color'], s=10, zorder=5)
    
    # 修改图例显示：所有条目统一为实线
    handles, labels = ax.get_legend_handles_labels()
    new_handles = []
    for h in handles:
        # 创建代理Artist，强制设置为实线（忽略实际线型）
        new_line = plt.Line2D([], [], 
                             color=h.get_color(), 
                             linestyle='-',  # 图例中强制实线
                             linewidth=2)
        new_handles.append(new_line)
    
    # 仅在第一个子图显示图例
    if ax_idx == 0:
        leg = ax.legend(handles=new_handles, labels=labels, 
                       loc='upper left', 
                       framealpha=0.3, 
                       fontsize=15)
        for text in leg.get_texts():
            text.set_fontname('Times New Roman')
    
    ax.tick_params(axis='both', which='major', labelsize=15)

####################
####################
####################
####################
####################
####################
#######################################################################################################################################################

# Fourth subplot: percentage change
ax = axes[3]
x = np.arange(len(files))
width = 0.18

# Stiffness types and colors
stiffness_types = {
    'Vertical Stiffness': '#4e79a7',
    'Lateral stiffness (LC1)': '#e15759',
    'Lateral stiffness (LC2)': '#59a14f'
}

# Calculate and plot percentage changes
for comp_idx, (comp_key, comp_config) in enumerate(components.items()):
    percentages = []
    for label, data in all_data.items():
        values = data[comp_key]
        percentage = ((values[-1] - values[0]) / values[0]) * 100
        percentages.append(percentage)
    
    # Get color based on component type
    if comp_key == 'strain_energy_g':
        color = stiffness_types['Vertical Stiffness']
    elif comp_key == 'strain_energy_l1':
        color = stiffness_types['Lateral stiffness (LC1)']
    else:
        color = stiffness_types['Lateral stiffness (LC2)']
    
    rects = ax.bar(x + comp_idx*width, percentages, width, 
                  color=color,
                  edgecolor='#333333',
                  linewidth=0.5,
                  alpha=0.85)
    
    # Add value labels
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.1f}%',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),
                    textcoords="offset points",
                    ha='center', 
                    va='bottom',
                    fontsize=15,
                    fontname='Times New Roman',
                    fontweight='bold')

# Add legend for stiffness types
legend_elements = [plt.Rectangle((0,0),1,1, color=color, label=label) 
                  for label, color in stiffness_types.items()]
ax.legend(handles=legend_elements, 
          loc='lower left', 
          framealpha=0.1,
          fontsize=8.6)

# Format fourth subplot
ax.set_xlabel(subplot_labels[3], fontsize=15, labelpad=10, fontname='Times New Roman')
ax.set_ylabel('Percentage Change (%)', fontsize=15, labelpad=3, fontname='Times New Roman')
ax.tick_params(axis='y', 
              labelsize=15,  # 字号
              labelcolor='black',  # 颜色
              pad=1) 
ax.set_xticks(x + width)
ax.set_xticklabels([config['label'] for config in files.values()], 
                  fontsize=12, 
                  rotation=0,
                  fontname='Times New Roman')

# Set font for tick labels
for label in ax.get_xticklabels() + ax.get_yticklabels():
    label.set_fontname('Times New Roman')

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.savefig('Moo_no_markers.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import json
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.lines import Line2D

# ==================== 全局设置 ====================
font_s = 15  # 基础字体大小
value_fontsize = 20  # 数值标记字体大小
NS = 20  # 标记点大小

plt.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Times New Roman'],
    'font.size': font_s,
    'axes.titlesize': 14,
    'axes.titleweight': 'bold',
    'axes.labelweight': 'bold',
    'xtick.labelsize': 14,
    'ytick.labelsize': 14,
    'legend.fontsize': 11,
    'axes.grid': False,
    'font.style': 'normal'
})

# ==================== 数据配置 ====================
# 全部数据文件配置
all_files = {
    'FEMoo_data.json': {'label': 'Combined FE', 'color': '#ff7f0e'}, 
    'LC1_data.json': {'label': 'LC1* OPT', 'color': '#59a14f'},
    'LC2_data.json': {'label': 'LC2* OPT', 'color': '#edc948'},
    'Moo_data.json': {'label': 'Combined Hybrid', 'color': '#9467bd'},
}

# 需要比较的两种方法（第四张图专用）
files_to_compare = {
    'FEMoo_data.json': all_files['FEMoo_data.json'],
    'Moo_data.json': all_files['Moo_data.json']
}

# 应变能分量配置
components = {
    'strain_energy_g': {'name': 'Gravity', 'unit': 'KJ'},
    'strain_energy_l1': {'name': 'Load Case 1', 'unit': 'KJ'},
    'strain_energy_l2': {'name': 'Load Case 2', 'unit': 'KJ'}
}

# ==================== 数据加载 ====================
all_data = {}
for file, config in all_files.items():
    try:
        with open(file) as f:
            data = json.load(f)
        total_iter = data['metadata']['parameters']['epochs']
        iterations_data = data['iterations']
        
        all_data[config['label']] = {
            'strain_energy_g': [iterations_data[i]['strain_energy_g'] for i in range(total_iter + 1)],
            'strain_energy_l1': [iterations_data[i]['strain_energy_l1'] for i in range(total_iter + 1)],
            'strain_energy_l2': [iterations_data[i]['strain_energy_l2'] for i in range(total_iter + 1)]
        }
    except FileNotFoundError:
        print(f"Error: {file} not found!")
    except KeyError as e:
        print(f"Error in {file}: Missing key {e}")

# ==================== 绘图设置 ====================
fig, axes = plt.subplots(2, 2, figsize=(12, 12))
fig.subplots_adjust(hspace=0.3, wspace=0.3)
axes = axes.ravel()
subplot_labels = ['(1)', '(2)', '(3)', '(4)']

# ==================== 前三个子图（全部数据） ====================
for ax_idx, (comp_key, comp_config) in enumerate(components.items()):
    ax = axes[ax_idx]
    first_curve = True
    
    for label, data in all_data.items():
        values = data[comp_key]
        normalized_values = [v / values[0] for v in values]
        
        # 获取当前曲线的配置
        config = next(cfg for cfg in all_files.values() if cfg['label'] == label)
        
        # 绘制曲线
        ax.plot(range(len(values)), normalized_values,
                color=config['color'],
                linestyle='-',
                linewidth=2,
                label=label)
        
        # 标记起点和终点
        ax.scatter(0, normalized_values[0], color=config['color'], s=NS, zorder=5)
        ax.scatter(len(values)-1, normalized_values[-1], color=config['color'], s=NS, zorder=5)
        
        # 标注起点数值（仅第一条曲线）
        if first_curve:
            ax.text(0, normalized_values[0], f"{normalized_values[0]:.2f}",
                   ha='left', va='bottom' if normalized_values[0] > 1 else 'top',
                   fontsize=value_fontsize, fontname='Times New Roman',
                   bbox=dict(facecolor='white', alpha=0.8, edgecolor='none'))
            first_curve = False
        
        # 标注终点数值
        ax.text(len(values)-1, normalized_values[-1], f"{normalized_values[-1]:.2f}",
               ha='right', va='bottom' if normalized_values[-1] > 1 else 'top',
               fontsize=value_fontsize, fontname='Times New Roman',
               bbox=dict(facecolor='white', alpha=0.8, edgecolor='none'))
    
    # 子图格式设置
    ax.set_xlabel(subplot_labels[ax_idx], fontsize=font_s, labelpad=10)
    ax.set_ylabel("Normalized Energy", fontsize=font_s, labelpad=8)
    
    # 设置刻度标签字体
    for label in ax.get_xticklabels() + ax.get_yticklabels():
        label.set_fontname('Times New Roman')
    
    # 仅第一个子图显示图例
    if ax_idx == 0:
        leg = ax.legend(loc='upper left', framealpha=0.3, fontsize=11)
        for text in leg.get_texts():
            text.set_fontname('Times New Roman')

# ==================== 第四个子图（两种方法比较） ====================
ax = axes[3]
width = 0.35  # 柱状图宽度
x = np.arange(len(files_to_compare))  # 两种方法的位置
case_colors = ['#4e79a7', '#e15759', '#59a14f']  # 工况颜色（Gravity/LC1/LC2）

for comp_idx, (comp_key, comp_config) in enumerate(components.items()):
    percentages = []
    for label in [config['label'] for config in files_to_compare.values()]:
        values = all_data[label][comp_key]
        percentage = ((values[-1] - values[0]) / values[0]) * 100
        percentages.append(percentage)
    
    # 绘制柱状图
    rects = ax.bar(x + comp_idx*width, percentages, width,
                  color=case_colors[comp_idx],
                  edgecolor='#333333',
                  linewidth=0.5,
                  alpha=0.85,
                  label=comp_config['name'])
    
    # 添加数值标签
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.1f}%',
                   xy=(rect.get_x() + rect.get_width()/2, height),
                   xytext=(0, 3),
                   textcoords="offset points",
                   ha='center', va='bottom',
                   fontsize=18,
                   fontname='Times New Roman',
                   fontweight='bold')

# 第四子图格式设置
ax.set_xlabel(subplot_labels[3], fontsize=font_s, labelpad=10)
ax.set_ylabel('Percentage Change (%)', fontsize=font_s, labelpad=8)
ax.set_xticks(x + width)
ax.set_xticklabels([config['label'] for config in files_to_compare.values()],
                  fontsize=11,
                  rotation=0,
                  fontname='Times New Roman')

# 添加工况图例
legend_elements = [Line2D([0], [0], color=case_colors[i], lw=4, label=list(components.values())[i]['name'])
                  for i in range(len(components))]
ax.legend(handles=legend_elements, loc='upper right', fontsize=9)

# ==================== 保存和显示 ====================
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.savefig('Final_Comparison.png', dpi=300, bbox_inches='tight')
plt.show()
```
